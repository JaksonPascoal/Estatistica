```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Criar um Data Frame de Exemplo
fat = [25,38,29,115,82, 114, 120, 160,153,190,239,175]
```

```python
# Meses
mes = list(range(1,13))
```

```python
mes
```

```python
# Criar um dataframe
data_dict = {'Mes': mes, 'Faturamento': fat}
```

```python
df = pd.DataFrame(data_dict)
```

```python
df.head()
```

```python
# Visualizar dados através do grafico de dispersão
x = df['Mes']
y = df['Faturamento']
plt.scatter(x,y)
plt.show;
```

Acima percebe-se uma correlação posiitiva dos dados

```python
#Verificar correlações
df[['Mes','Faturamento']].corr()
```

```python
# importar biblioteca sklearn
from sklearn import linear_model
import numpy as np
```

```python
#instanciar modelo de Regressoa Linear
lm = linear_model.LinearRegression()
```

```python
#Reshape x (remodelar x) Se for somente uma variavel é preciso remodelar "x"
x = df[['Mes']]
y = df[['Faturamento']]
```

```python
x = np.array(x).reshape(-1,1)
```

```python
# Treinar Modelo
sk_modelo = lm.fit(x,y)
```

```python
# Gerar Predições
skpredicoes = sk_modelo.predict(x)
```

```python
skpredicoes
```

```python
lm.score(x,y)
```

```python
lm.coef_
```

```python
lm.intercept_
```
